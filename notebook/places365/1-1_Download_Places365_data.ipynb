## 2. CNN models

[[$\leftarrow$ Prev. part]](#1.-Places365-dataset) $\qquad$ [[Back to top]](#Navigation)

For this dataset, there are 2 pretrained CNN models relevant to our study:
- [Alexnet](https://pytorch.org/hub/pytorch_vision_alexnet/)
- [ResNet18](https://pytorch.org/hub/pytorch_vision_resnet/)

Available here:
- https://github.com/CSAILVision/places365

````python
# alexnet_places365.pth.tar
# resnet18_places365.pth.tar

model_urls: dict[str, str] = {
    "alexnet": "http://places2.csail.mit.edu/models_places365/alexnet_places365.pth.tar",
    "resnet18": "http://places2.csail.mit.edu/models_places365/resnet18_places365.pth.tar"
}
````

````python
os.makedirs(model_dirname_path, exist_ok=True)
model_name: str
model_url: str
for model_name, model_url in model_urls.items():
    current_path = os.path.join(model_dirname_path, model_url.split('/')[-1])
    ! wget {model_url} -O {current_path}
    print(f'Model {model_name} was successfully downloaded to {current_path}')
````
